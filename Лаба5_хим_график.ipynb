```python
import matplotlib.pyplot as plt

x = [0.01, 0.1, 1.0]
y1 = [150.0, 81.8, 53.1]
y2 = [41, 70.5, 100]

fig, ax = plt.subplots(figsize=(14, 10))

ax.plot(
    x,
    y1,
    color="black",
    marker="s",
    markersize=10,
    linewidth=3,
    linestyle="-",
    label="Эмпирические значения",
)

ax.plot(
    x,
    y2,
    color="black",
    marker="^",
    markersize=9,
    linewidth=1.5,
    linestyle="--",
    label="$Теоретические значения (для T_{реакц.} = 298 К)$",
)

ax.set_xscale("log")
ax.set_xlabel(r"$c(CuSO_4), M$", fontsize=20, color="black")
ax.set_ylabel(r"$U, мВ$", fontsize=20, color="black")

ax.tick_params(colors="black", which="both", labelsize=10)
for spine in ax.spines.values():
    spine.set_color("black")

ax.grid(True, which="both", linestyle=":", linewidth=0.5, alpha=0.6)
ax.legend(frameon=True, fontsize='x-large')

plt.tight_layout()
plt.show()
```
