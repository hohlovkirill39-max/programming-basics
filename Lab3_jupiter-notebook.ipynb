Задача 5. Создайте датафрейм, в котором будет среднее количество звонков Total day calls для каждого штата.

```python
import pandas as pd
d = pd.read_csv('telecom_churn.csv')
p = d.groupby('State')['Total day calls'].mean().reset_index(name='Total day calls')
p
```

Задача 6. Оставьте в созданном датафрейме строки только с теми штатами, где количество звонков Total day calls больше среднего по исходному датафрейму.

```python
import pandas as pd
d = pd.read_csv('telecom_churn.csv')
i = d['Total day calls'].mean()
p = d.groupby('State')['Total day calls'].mean().reset_index(name='Total day calls')
r = p.loc[(p['Total day calls'] > i)]
r
```

Задача 7. Создайте датафрейм, в котором будет средние количества звонков Total day calls и Total eve calls для каждого штата.

```python
import pandas as pd
d = pd.read_csv('telecom_churn.csv')
d.groupby('State')[['Total day calls', 'Total eve calls']].mean()
```

Задача 8. Создайте датафрейм, в котором будует средние количества звонков Total day calls и Total eve calls для каждого штата, а также столбец со значениями True и False - ответом на вопрос, больше ли дневных звонков, чем вечерних.

```python
import pandas as pd
d = pd.read_csv('telecom_churn.csv')
m = d.groupby('State')[['Total day calls', 'Total eve calls']].mean()
m['Day > Eve'] = m['Total day calls'] > m['Total eve calls']
m
```
