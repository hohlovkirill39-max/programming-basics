Задание №0.

Построение графиков функций:
1. $f_1(x) = \sin(x)$;
2. $f_2(x) = x$;
3. $f_3(x) = \sec(\sin(x))$
4. $f_4(x) = x^{\sin(x)}$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 10*np.pi, 0.01)
plt.figure(figsize=(5, 5))
plt.plot(x, np.sin(x), 'red', label=r'$\sin(x)$')
plt.plot(x, x, 'black', label=r'$x$')
plt.plot(x, 1/(np.cos(np.sin(x))), 'blue', label=r'$\sec(\sin(x))$')
plt.plot(x, x**np.sin(x), 'green', label=r'$x^{\sin(x)}$')
plt.legend()
plt.axis('equal')
plt.grid()
plt.show()
```

Задание №3

Построение графика функции $\log_{1+\tan\left(\frac{1}{1+{\sin^2}(x)}\right)}(x^2 + 1)\exp{\left(-\frac{|x|}{10}\right)}$.

```python
import numpy as np
import math
import matplotlib.pyplot as plt

x = np.arange(-50, 50, 0.01)
a = 1 + np.tan(1 / (1+(np.sin(x)**2)))
c = x**2 + 1
f = -np.abs(x)/10
d = np.exp(f)
b = np.log(c*d) / np.log(a)
plt.figure(figsize=(5, 5))
plt.plot(x, b, 'purple', label=r'$f(x) = \log_{1+\tan\left(\frac{1}{1+{\sin^2}(x)}\right)}(x^2 + 1)\exp{\left(-\frac{|x|}{10}\right)}$')
plt.legend()
plt.grid()
plt.show()
```

Задание №6*

Построение графика функции Вейерштрасса:
$w(x) = \sum_{n=0}^\infty b^n \cos(a^n \pi x)$.

```python
import numpy as np
import matplotlib.pyplot as plt

a = 3 #a=2k+1, a≠1
b = 0.5 #0<b<1
x = np.arange(-2, 2, 0.01)
w = 0 * x
for n in range(10):
    w += b**n * np.cos(a**n * np.pi * x)

plt.figure(figsize=(5, 5))
plt.plot(x, w, 'purple')
plt.grid()
plt.show()
```
